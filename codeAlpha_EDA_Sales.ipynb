```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df['Sales'],bins=30)
plt.xlabel('Sales')
plt.ylabel('Number of Orders')
plt.title('Distribution of sales')
plt.show()
```

```python
category_sales=df.groupby('Category')['Sales'].sum()
category_sales
```

```python
category_sales.plot(kind='bar',figsize=(8,5))

plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.title('Total Sales by Category')
plt.xticks(rotation=0)
plt.show()
```

```python
region_sales=df.groupby('Region')['Sales'].sum()
region_sales
```

```python
region_sales.plot(kind='bar',figsize=(8,5))

plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.title('Total Sales by Region')
plt.xticks(rotation=0)
plt.show()
```

```python
segment_sales=df.groupby('Segment')['Sales'].sum()
segment_sales
```

```python
segment_sales.plot(kind='bar',figsize=(8,5))

plt.xlabel('Segment')
plt.ylabel('Total Sales')
plt.title('Total Sales by Segment')
plt.xticks(rotation=0)
plt.show()
```

## Key Insights

1. The dataset contains 9,800 records and 18 columns.
2. Only the Postal Code column contains missing values (11 records).
3. Technology has the highest total sales among the three categories.
4. The sales distribution is right-skewed, with a few very high sales values.
5. The West region has the highest total sales, while the South region has the lowest.
6. The Consumer segment has the highest total sales among the three segments.
